```python
from nltk.corpus import stopwords

presidents = {
    "Abraham Lincoln": "government people nation freedom government people union freedom",
    "Barack Obama": "people change hope nation people change future hope",
    "Donald Trump": "america people great country america great people country",
    "Joe Biden": "people democracy nation unity people democracy future unity"
}

stop_words = set(stopwords.words('english'))

for president, text in presidents.items():
  words = nltk.word_tokenize(text.lower(), preserve_line=True)
  words = [w for w in words if w.isalpha() and w not in stop_words]

    # Word frequencies
  freq = FreqDist(words)
  print("\n", president, ":", freq.most_common(5))

    # Word cloud
  cloud = WordCloud(width=600, height=300,
                      background_color='white').generate(' '.join(words))

  plt.figure(figsize=(7, 3))
  plt.imshow(cloud, interpolation='bilinear')
  plt.axis('off')
  plt.title(president)
  plt.show()
```
